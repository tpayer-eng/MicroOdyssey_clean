# Beta diversity: PERMANOVA and PERMIDSP

**Goal:** This notebook inspects whether microbial community composition differs across infant age groups using Bray–Curtis dissimilarities.

PERMANOVA tests differences between groups, whereas PERMDISP assesses differences in within-group spread.

## 0. Setup

```python
from qiime2 import Visualization

%cd /home/jovyan/MicroOdyssey_clean

data_dir = "output/04_diversityanalyses/core-metrics-results"
metadata_dir = "data"
result_dir = "output/04_diversityanalyses"
```

## 1. PERMANOVA

PERMANOVA was computed to test differences in microbial community composition across infant age groups using Bray–Curtis dissimilarities and 999 permutations.

```python
! qiime diversity beta-group-significance \
      --i-distance-matrix $data_dir/bray_curtis_distance_matrix.qza \
      --m-metadata-file $metadata_dir/metadata_heatmap.tsv \
      --m-metadata-column age_group_ordered \
      --p-method permanova \
      --p-permutations 999 \
      --o-visualization $result_dir/braycurtis_permanova.qzv 
```

```python
Visualization.load(f"{result_dir}/braycurtis_permanova.qzv")
```

## 2. PERMDISP

```python
! qiime diversity beta-group-significance \
      --i-distance-matrix $data_dir/bray_curtis_distance_matrix.qza \
      --m-metadata-file $metadata_dir/metadata_heatmap.tsv \
      --m-metadata-column age_group_ordered \
      --p-method permdisp \
      --p-permutations 999 \
      --o-visualization $result_dir/braycurtis_permdisp.qzv 
```

```python
Visualization.load(f"{result_dir}/braycurtis_permdisp.qzv")
```
